# Coleta de Dados de Landmarks das Mãos

Este notebook captura os marcos (landmarks) das mãos usando o MediaPipe e salva as coordenadas (x, y, z) em um arquivo CSV junto com um selo (label).

### Instruções:
1. Defina a `LABEL` que deseja coletar no bloco de código abaixo.
2. Execute o código.
3. Pressione `s` no teclado para salvar o frame atual (as landmarks da primeira mão detectada) no CSV.
4. Pressione `q` para sair e fechar o arquivo.

```python
import cv2
import mediapipe as mp
import numpy as np
import os
import csv

# --- CONFIGURAÇÕES ---
LABEL = "va se foder"  # Mude aqui para a classe que está coletando
CSV_FILE = "hand_landmarks_data.csv"
MP_MODEL_PATH = "gesture_recognizer.task"

# Inicializa utilitários de desenho
mp_hands_connections = mp.tasks.vision.HandLandmarksConnections
mp_drawing = mp.tasks.vision.drawing_utils
mp_drawing_styles = mp.tasks.vision.drawing_styles

def save_to_csv(landmarks, label, filename):
    """Salva as coordenadas x, y, z de 21 landmarks + label em um CSV."""
    file_exists = os.path.isfile(filename)
    
    with open(filename, mode='a', newline='') as f:
        writer = csv.writer(f)
        
        # Se o arquivo não existir, escreve o cabeçalho
        if not file_exists:
            header = []
            for i in range(21):
                header.extend([f'x{i}', f'y{i}', f'z{i}'])
            header.append('label')
            writer.writerow(header)
        
        # Prepara a linha de dados
        data_row = []
        for lm in landmarks:
            data_row.extend([lm.x, lm.y, lm.z])
        data_row.append(label)
        
        writer.writerow(data_row)

def main():
    if not os.path.exists(MP_MODEL_PATH):
        print(f"Erro: O arquivo de modelo {MP_MODEL_PATH} não foi encontrado.")
        return

    # Configuração do MediaPipe Tasks
    BaseOptions = mp.tasks.BaseOptions
    GestureRecognizer = mp.tasks.vision.GestureRecognizer
    GestureRecognizerOptions = mp.tasks.vision.GestureRecognizerOptions
    VisionRunningMode = mp.tasks.vision.RunningMode

    options = GestureRecognizerOptions(
        base_options=BaseOptions(model_asset_path=MP_MODEL_PATH),
        running_mode=VisionRunningMode.VIDEO,
        num_hands=1, # Coletando apenas de uma mão por simplicidade
        min_hand_detection_confidence=0.5,
        min_hand_presence_confidence=0.5,
        min_tracking_confidence=0.5,
    )

    cap = cv2.VideoCapture(0)
    print(f"\nColetando dados para a label: '{LABEL}'")
    print("Pressione 's' para salvar landmarks ou 'q' para sair.")

    count = 0
    with GestureRecognizer.create_from_options(options) as recognizer:
        while cap.isOpened():
            success, frame = cap.read()
            if not success: break

            frame = cv2.flip(frame, 1)
            rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            mp_image = mp.Image(image_format=mp.ImageFormat.SRGB, data=rgb_frame)
            timestamp_ms = int(cv2.getTickCount() / cv2.getTickFrequency() * 1000)
            
            # Extração
            result = recognizer.recognize_for_video(mp_image, timestamp_ms)

            current_landmarks = None
            if result.hand_landmarks:
                current_landmarks = result.hand_landmarks[0]
                # Desenha landmarks
                mp_drawing.draw_landmarks(
                    frame,
                    current_landmarks,
                    mp_hands_connections.HAND_CONNECTIONS,
                    mp_drawing_styles.get_default_hand_landmarks_style(),
                    mp_drawing_styles.get_default_hand_connections_style()
                )

            # UI
            cv2.putText(frame, f"Label: {LABEL} | Capturados: {count}", (10, 30), 
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)
            
            cv2.imshow('Coleta de Dados - Landmark Recording', frame)

            key = cv2.waitKey(1) & 0xFF
            if key == ord('q'):
                break
            elif key == ord('s'):
                if current_landmarks:
                    save_to_csv(current_landmarks, LABEL, CSV_FILE)
                    count += 1
                    print(f"[{count}] Frame salvo para '{LABEL}'")
                else:
                    print("Nenhuma mão detectada para salvar.")

    cap.release()
    cv2.destroyAllWindows()
    print(f"\nProcesso finalizado. Total de capturas: {count}")
    print(f"Dados salvos em: {CSV_FILE}")

if __name__ == "__main__":
    main()
```
